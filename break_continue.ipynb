### Break and Continue statements

```python
sum = 10

for i in range(1, 11):
    sum = sum + i
    print(i, "is added to the sum")
```

```python
sum = 0

for i in range(1, 11):
    sum+= i
    print(i, "is added to the sum")

print(sum)
```

```python
sum = 0

for i in range(1, 11):

    print (i, "is processing")

    if i % 2 == 0: ##even
        continue

    sum += i
    print(i, "is added to the sum")

print(sum)
```

```python
accuracy = 30

for i in range(100):
    accuracy += 1
    print(accuracy)
    if accuracy == 100 :
        break
```
