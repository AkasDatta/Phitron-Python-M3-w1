### Formated String

```python
name = "Adil"
age = 23
height = 5.3456

print(f"His name is {name}. He is {age} years old. He is {height} feel tall. \n aa")
```

```python
name = "Adil"
age = 23
height = 5.3456

info = f"His name is {name.upper()}. He is {age} years old. He is {height:.2} feel tall. \n aa"

print(info)
print(type(info))

model_accuracy = 0.8234

print("the model accuracy is {model_accuracy}")
print(f"the model accuracy is {model_accuracy}")
print(f"the model accuracy is {model_accuracy:.2%}")
print(f"the model accuracy is {model_accuracy+10}")
```
