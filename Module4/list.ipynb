### Python list
###
### python এর list হচ্ছে একটা কন্টেইনার যার মধ্যে একের অধিক ভ্যালু স্টোর করে রাখা যায়। অর্থাৎ list হচ্ছে collection of data.

```python
## declaration part
numbers = [100, 20, 34, 56, 78, 100, 100]
float_numbers = [10.23, 23.34, 45.4, 33.56, 31.23]
fruits = ['apple', 'banana', 'grapes', 'quicumber']
mix = [23, 'banana', 10.2, True]

list_er_moddhe_list = [[100, 20, 23],[10, 20, 23],[100, 40, 23]]

print(type(numbers))
print(float_numbers)
print(type(numbers))
print(fruits)
print(type(fruits[0]))
print(mix)
print(type(mix))
print(type(mix[3]))
print(list_er_moddhe_list)
print(list_er_moddhe_list[1])
print(list_er_moddhe_list[1][2])
```

```python
## access

print(numbers[2])

numbers[3] = 89

print(numbers[3])

new_list = numbers[0:3]
print(new_list)
```

```python
## modification
## adding a item

numbers.append(80)
print(numbers)

# insert

numbers.insert(0,45)
print(numbers)


# deleting (deleting last numbers)

numbers.pop()
print(numbers)

# remove (deleting a specfic value)
numbers.remove(100)
print(numbers)
```

```python
##sort function

numbers.sort()
print(numbers)

numbers.reverse()
print(numbers)

# reverse sort 

numbers.sort(reverse=True)
print(numbers)
```

### stack (last in last out)

```python
stack = []

stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)

print(stack)
```

```python
stack = []

stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)

# top element
print(f"top element : {stack[-1]}")
# getting the top and last element 
print(stack.pop())
print(stack.pop())
print(stack.pop())
print(stack.pop())
print(stack.pop())
```

### Queue last in first out

```python
queue = []

# insertion 
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

# getting the top and last element 
print(queue.pop(0))
print(f"front e {queue[0]}")
print(queue.pop(0))
print(f"front e {queue[0]}")
print(queue.pop())
print(queue.pop())
print(queue.pop())
```

### List comprehension

```python
even = []

for i in range(100):
    if i%2==0:
        even.append(i)

        print(even)
```

```python
even = []
## naive way
for i in range(100):
    if i%2==0:
        even.append(i)

print(even)
```
