### Python list
###
### python এর list হচ্ছে একটা কন্টেইনার যার মধ্যে একের অধিক ভ্যালু স্টোর করে রাখা যায়। অর্থাৎ list হচ্ছে collection of data.

```python
## declaration part
numbers = [100, 20, 34, 56, 78, 100, 100]

print(type(numbers))
```

```python
## access

print(numbers[2])

numbers[3] = 89

print(numbers[3])

new_list = numbers[0:3]
print(new_list)
```

```python
## modification
## adding a item

numbers.append(80)
print(numbers)

# insert

numbers.insert(0,45)
print(numbers)


# deleting (deleting last numbers)

numbers.pop()
print(numbers)

# remove (deleting a specfic value)
numbers.remove(100)
print(numbers)
```
