```python
# cleaning process 

from functools import reduce

with open("./sample_data/sample.txt", "r") as file:
    strings_list = file.readlines()

    print(strings_list)

# total line koeta 
    total_lines = len(strings_list)
    print(total_lines)

    # word ber korbo 
    number_of_words = list(map(lambda x : len(x.split()),strings_list))
    print(number_of_words)

# total words
    total_number_of_words = reduce(lambda x,y : x+y , number_of_words)
    print(total_number_of_words)

    # cleaning process (new line delete)
    strings_list = list(map(str.strip,strings_list))
    print(strings_list)

    # space delete 
    strings_list = list(map(lambda x : x.replace(" ",""), strings_list))
    print(strings_list)

    # koeta character ache 

    number_of_character = list(map(lambda x : len(x), strings_list))
    print(number_of_character)

    # total number of chareacter 
    total_number_of_character = reduce(lambda x,y : x+y, number_of_character)
    print(total_number_of_character)
```

```python
with open("./sample_data/counter_of_string.txt", "w") as file:
    file.write(f"total line: {total_lines}\n total number of words:{total_number_of_words}\n total number of characters :{total_number_of_character}")
```
