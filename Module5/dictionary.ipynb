### Dictionary methods

```python
dic = { "name" : "Python" , "age" : 20 , "address" : "Feni" , "numbers" : [10, 20, 30]}


# printing default value
print(dic.get("math_marks"))
print(dic.get("math_marks",0))
```

```python
# adding a value 
dic = { "name" : "Python" , "age" : 20 , "address" : "Feni" , "numbers" : [10, 20, 30]}


# printing default value
print(dic)
dic["math_marks"] = 29
print(dic)


dic.update({"engliah_marks": 34})
print(dic)
```

```python
#deleteing

del dic["age"]

print(dic)
```

```python
# copy 
dic_2 = dic.copy()
print(dic_2)
```

```python
#  testing 

dic_3 = {(10, 20, 34) : "a"}
print(dic_3)

dic_3 = {"key" : (10, 20, 34)}
print(dic_3)
dic_3 = {"key" : {"name" : "adil", "age" : 29}}
print(dic_3)
```

```python
dic = { "name" : "Python" , "age" : 20 , "address" : "Feni" , "numbers" : [10, 20, 30]}

keys = dic.keys()
print(keys)

dic["math_numbers"] = 20
print(keys)

values = dic.values()
print(values)

dic["math_numbers"] = 40
print(values)
```

```python
# items
items = dic.items()
print(items)

# iteration 
for key,values in dic.items():
    print(key,values)
```
