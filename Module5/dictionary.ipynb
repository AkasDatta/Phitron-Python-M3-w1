### Dictionary methods

```python
dic = { "name" : "Python" , "age" : 20 , "address" : "Feni" , "numbers" : [10, 20, 30]}


# printing default value
print(dic.get("math_marks"))
print(dic.get("math_marks",0))
```

```python
# adding a value 
dic = { "name" : "Python" , "age" : 20 , "address" : "Feni" , "numbers" : [10, 20, 30]}


# printing default value
print(dic)
dic["math_marks"] = 29
print(dic)


dic.update({"engliah_marks": 34})
print(dic)
```

```python
#deleteing

del dic["age"]

print(dic)
```

```python
# copy 
dic_2 = dic.copy()
print(dic_2)
```

```python
#  testing 

dic_3 = {(10, 20, 34) : "a"}
print(dic_3)

dic_3 = {"key" : (10, 20, 34)}
print(dic_3)
dic_3 = {"key" : {"name" : "adil", "age" : 29}}
print(dic_3)
```

```python
dic = { "name" : "Python" , "age" : 20 , "address" : "Feni" , "numbers" : [10, 20, 30]}

keys = dic.keys()
print(keys)

dic["math_numbers"] = 20
print(keys)

values = dic.values()
print(values)

dic["math_numbers"] = 40
print(values)
```

```python
# items
items = dic.items()
print(items)

# iteration 
for key,values in dic.items():
    print(key,values)
```

### dictionary compression

```python
print(list(range(10)))

square = {x:x**2 for x in range(10)}
print(square)
square = {x:x**2 for x in range(1, 10) if x%2==0}
print(square)
square = {x:x**2 for x in range(1, 10) if x%3==0}
print(square)
```

```python
co_ordinates = [(10, 20, 30), (20, 3, 7), (11, 23)]
locations = ["dhaka", "chattagram", "sylhet"]

exect_locations = {co_or : loc for co_or, loc in zip(co_ordinates, locations)}
print(exect_locations)
```

```python
lst = [10, 20, 10, 20, 30, 50, 66, 38]

unique_values = set(lst)
print(unique_values)

lst = list(unique_values)
print(lst)
```

```python
string = """You are" is a fundamental English phrase combining the second-person pronoun "you" with the present tense form of the verb "to be" ("are")"""

words = string.split()
count = {}

for word in words:
    print(count.get(word))
    print(count.get(word,0))
    count[word] = count.get(word,0) +1

print(count)

for k,v in count.items():
    print(f"count of {k} is {v}")
```
