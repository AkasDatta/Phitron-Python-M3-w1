```python
S = {1, 2, 3, 4, 12, 4, 6}

if (3 in S):
    print("Is present")
else:
    print("Not Present")
```

```python
S = {1, 2, 3, 4, 12, 4, 6}

if (3 in S):
    print("Is present")
else:
    print("Not Present")

for element in S:
    print(element)
```

```python
S = {1, 2, 3, 4, 12, 4, 6}

if (3 in S):
    print("Is present")
else:
    print("Not Present")

sum = 0

for element in S:
    sum += element

print(sum)
```

```python
# mutable 
s = {1, 3, 4, 2, 5, 6, 7, 8}

s.add(20)
s.add(0)

print(s)

s.pop()
s.pop()
print(s)

s.remove(2)
print(s)
```

### Methods of set

### Set mathematical

```python
#Union set
a = {1, 2, 3, 4}
b = {5, 2, 3, 8}

print(a.union(b))
```

```python
#intersection, disjoint, subset set
a = {1, 2, 3, 4}
b = {5, 2, 3, 8}

print(a.intersection(b))
print(a.isdisjoint(b))
print(a.issubset(b))
```
