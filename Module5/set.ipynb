```python
S = {1, 2, 3, 4, 12, 4, 6}

if (3 in S):
    print("Is present")
else:
    print("Not Present")
```

```python
S = {1, 2, 3, 4, 12, 4, 6}

if (3 in S):
    print("Is present")
else:
    print("Not Present")

for element in S:
    print(element)
```

```python
S = {1, 2, 3, 4, 12, 4, 6}

if (3 in S):
    print("Is present")
else:
    print("Not Present")

sum = 0

for element in S:
    sum += element

print(sum)
```
