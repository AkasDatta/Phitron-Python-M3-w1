### 1. collection of data

```python
# declaration

tup = (10, 20, 30, 40)
print(type(tup), tup)


float_tup = (10, 5, 0, 2)
print(type(float_tup), float_tup)

mixed_tup = (10, 10.5, "str", True)

print(mixed_tup)\

tup1 = (10)

tup2 = (10, 20)
tup3 = tuple([10])
tup3 = tuple([10,])
print(type(tup1), tup1)
print(type(tup2), tup2)
print(type(tup3), tup3)
```

```python
#access
tup = (10, 20, 10.2, True)

# index
print(tup)
print(tup[3])


# slicing 

new_tup = tup[0:3]
print(new_tup)
```

```python
#mutability

lst = [10, 2, 20, 34]
lst.append(60)
lst[1] = 100
print(lst)


# immutability
tup = (10, 20, 30, 23)
tup[1] = 100
print(tup)
```

```python
#tuple method
tup = (10, 20, 30, 23, 10, 20, 10)

print(tup.count(10))
print(tup.index(10))
```
