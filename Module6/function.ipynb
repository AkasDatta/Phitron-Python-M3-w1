# python functions 

### input > function >output

```python
# no parameter , no return type function

def welcome_message():
    print("Hello user")
    
welcome_message()
```

```python
def welcome_message(user):
    print(f"hello {user}")

welcome_message("python")
```

```python
def welcome_message(user="none"):
    print(f"hello {user}")

welcome_message()
```

```python
def adder (a, b):
    """
    add two numbers
    """
    sum = a + b
    print(sum)

adder(12, 34)

help(adder)
```

```python
def adder (a, b):
    sum = a + b * 5
    print(sum)

adder(12, 34)
```

```python
# positional argument 

def adder(a, b, c):
    print(f"{a}, {b}, {c}")


adder(10, 70, 56)
```

```python
#  argument 

def adder(* args):
    for i in args:
        print(i)


adder(10, 70, 56, 56)
```

```python
# keyword arguments 
def adder(**kwords):
    print(kwords)

adder( a= 19, b = 23, c = 23)
```

```python
# keyword arguments 
def adder(**kwords):
    for key,value in kwords.items():
        print(f"{key}: {value}")

adder( a= 19, b = 2333, c = 23)
```
