```python
# no parameter , no return type function

def welcome_message():
    print("Hello user")
    
welcome_message()
```

```python
def welcome_message(user):
    print(f"hello {user}")

welcome_message("python")
```

```python
def welcome_message(user="none"):
    print(f"hello {user}")

welcome_message()
```

```python
def adder (a, b):
    """
    add two numbers
    """
    sum = a + b
    print(sum)

adder(12, 34)

help(adder)
```

```python
def adder (a, b):
    sum = a + b * 5
    print(sum)

adder(12, 34)
```
